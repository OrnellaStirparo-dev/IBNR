```python
# 01_exploration.ipynb

# 🧩 Step 1. Imports
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 🧩 Step 2. Load data
df = pd.read_csv("../data/raw/claims_triangle.csv")
df.head()

# 🧩 Step 3. Explore structure
print(df.info())
print(df.describe())

# 🧩 Step 4. Basic visualization
plt.figure(figsize=(8,5))
sns.lineplot(data=df, x="development_year", y="cumulative_claims", hue="accident_year")
plt.title("Cumulative Claims by Development Year")
plt.show()

# 🧩 Step 5. Save processed version
df.to_csv("../data/processed/claims_triangle_clean.csv", index=False)
```
